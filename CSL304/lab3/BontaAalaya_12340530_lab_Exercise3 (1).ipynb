```python
import random, math

def conflicts(board):
    n = len(board)
    cnt = 0
    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == j - i:
                cnt += 1
    return cnt

def anneal(n=8, steps=5000, temp=50.0, cooling=0.995):
    state = [random.randint(0, n - 1) for _ in range(n)]
    best = state[:]
    best_conf = conflicts(state)
    curr_conf = best_conf
    t = temp
    
    for step in range(steps):
        new_state = state[:]
        if random.random() < 0.5:
            col = random.randrange(n)
            new_state[col] = random.randint(0, n - 1)
        else:
            c1, c2 = random.sample(range(n), 2)
            new_state[c1], new_state[c2] = new_state[c2], new_state[c1]

        new_conf = conflicts(new_state)
        delta = curr_conf - new_conf
        
        if delta > 0 or random.random() < math.exp(delta / t):
            state, curr_conf = new_state, new_conf

        if curr_conf < best_conf:
            best, best_conf = state[:], curr_conf

        t *= cooling
        if best_conf == 0:
            break

    return best, best_conf == 0

if __name__ == "__main__":
    solution, ok = anneal()
    if ok:
        print("solution found ", solution)
    else:
        print("no perfect solution, best attempt ", solution)
```
